```python
import kagglehub
import polars as pl
from pathlib import Path
```

```python
datapath = kagglehub.dataset_download("thec03u5/seinfeld-chronicles")
datapath
```

```python
scripts = pl.scan_csv(Path(datapath, "scripts.csv")).with_columns([
    pl.col("EpisodeNo").cast(pl.Int8),
    pl.col("Season").cast(pl.Int8)
]).collect()
scripts
```

```python
episode_info = pl.scan_csv(Path(datapath, "episode_info.csv")).with_columns([
    pl.col("EpisodeNo").cast(pl.Int8),
]).collect()
episode_info.tail()
```

```python
df = episode_info.join(scripts, left_on="SEID", right_on="SEID", how="left").select(["EpisodeNo", "Title", "SEID", "Character", "Dialogue"])
df
```
